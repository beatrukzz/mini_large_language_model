```python
# Dot product: multiply corresponding values and add the products

a = torch.tensor([1., 2., 3.])
b = torch.tensor([4., 5., 6.])

manual_dot_product = (1*4) + (2*5) + (3*6)
pytorch_dot_product = torch.dot(a, b)

print("Manual:", manual_dot_product)
print("PyTorch:", pytorch_dot_product)
```

```python
# Matrix multiplication shape rule:
# (m x n) @ (n x p) -> (m x p)

A = torch.tensor([[1., 2.],
                  [3., 4.],
                  [5., 6.]])

B = torch.tensor([[7., 8., 9.],
                  [10., 11., 12.]])

print("A shape:", A.shape)
print("B shape:", B.shape)

C = A @ B

print("Result shape:", C.shape)
print(C)
```

```python
# Matrix multiplication also requires compatible dtypes.

int_tensor = torch.tensor([[1, 2],
                           [3, 4]])

float_tensor = torch.tensor([[1., 2.],
                             [3., 4.]])

print(int_tensor.dtype)
print(float_tensor.dtype)

# Convert the integer tensor before matrix multiplication
int_tensor = int_tensor.to(torch.float32)

result = int_tensor @ float_tensor

print(result)
```
